# Data Preparation

## Objective

This notebook prepares the cleaned dataset for machine learning by separating features and target labels, encoding categorical crop labels into numerical values, and splitting the dataset into training and testing sets.

```python
# Import Required Libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
```

```python
# Load the cleaned dataset

df = pd.read_csv("../data/processed/crop_cleaned.csv")

df.head()
```

## Separate Features and Target

The input variables are stored in **X**, while the crop label is stored in **y**.

```python
# Separate independent and dependent variables

X = df.drop("label", axis=1)
y = df["label"]

print("Feature Shape :", X.shape)
print("Target Shape  :", y.shape)
```

## Encode Crop Labels

Machine learning algorithms work with numerical values. Therefore, each crop name is mapped to a unique integer.

```python
# Create label mappings

unique_labels = sorted(y.unique())

label_to_index = {
    label: index
    for index, label in enumerate(unique_labels)
}

index_to_label = {
    index: label
    for label, index in label_to_index.items()
}

label_to_index
```

```python
# Encode target labels

y_encoded = y.map(label_to_index)

y_encoded.head()
```

```python
# Check encoded labels

print("Number of Classes:", len(unique_labels))
print(unique_labels)
```

## Split the Dataset

The dataset is divided into training and testing sets using an 80:20 ratio. Stratified sampling is used to preserve the class distribution.

```python
# Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)
```

```python
# Display split sizes

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

```python
# Verify class distribution

print("Training Distribution")
print(y_train.value_counts().sort_index())

print()

print("Testing Distribution")
print(y_test.value_counts().sort_index())
```

```python
# Save label mappings for future prediction

import json

with open("../data/processed/label_mapping.json", "w") as file:
    json.dump(label_to_index, file, indent=4)

print("Label mapping saved successfully.")
```

```python
# Save train-test datasets

X_train.to_csv("../data/processed/X_train.csv", index=False)
X_test.to_csv("../data/processed/X_test.csv", index=False)

y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)

print("Training and testing datasets saved successfully.")
```

## Conclusion

The dataset has been successfully prepared for machine learning. Features and target variables were separated, crop labels were encoded into numerical values, and the dataset was split into training and testing sets. The next step is to implement the **Min-Max Scaler from scratch**.
