# Min-Max Scaling (From Scratch)

## Objective

The Min-Max Scaler normalizes numerical features to a common range between **0 and 1**. This improves model performance by ensuring that all features contribute equally during training.

The scaling formula is:

\[
X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}
\]

```python
# Import Required Libraries

import pandas as pd
import numpy as np
import pickle
```

```python
# Load Training and Testing Data

X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)
```

## Implement Min-Max Scaler

The scaler is implemented from scratch without using any machine learning libraries.

```python
class MinMaxScalerScratch:

    def __init__(self):
        self.min = None
        self.max = None

    def fit(self, X):
        """
        Compute the minimum and maximum value of each feature.
        """
        self.min = X.min()
        self.max = X.max()

    def transform(self, X):
        """
        Scale the features using the fitted minimum and maximum values.
        """
        denominator = self.max - self.min

        # Prevent division by zero
        denominator = denominator.replace(0, 1)

        X_scaled = (X - self.min) / denominator

        return X_scaled

    def fit_transform(self, X):
        """
        Fit the scaler and transform the data.
        """
        self.fit(X)
        return self.transform(X)
```

## Fit the Scaler on Training Data

```python
scaler = MinMaxScalerScratch()

X_train_scaled = scaler.fit_transform(X_train)
```

## Transform Testing Data

The testing data is transformed using the minimum and maximum values calculated from the training data.

```python
X_test_scaled = scaler.transform(X_test)
```

## Save Scaled Data

The scaled datasets are saved for model training.

```python
X_train_scaled.to_csv(
    "../data/processed/X_train_scaled.csv",
    index=False
)

X_test_scaled.to_csv(
    "../data/processed/X_test_scaled.csv",
    index=False
)

print("Scaled datasets saved successfully.")
```

```python
scaler_data = {
    "min": scaler.min,
    "max": scaler.max
}

with open("../saved_models/minmax_scaler.pkl", "wb") as file:
    pickle.dump(scaler_data, file)

print("Min-Max Scaler saved successfully.")
```
